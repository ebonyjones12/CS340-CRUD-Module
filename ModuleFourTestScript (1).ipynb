```python
# Python Testing Script

# Import the CRUD module
from CRUD_Python_Module import AnimalShelter

# Instantiate an instance of your class
shelter = AnimalShelter()
print("Successfully connected to the database!")

# Test CREATE - Use your create function to create a new record in the aac database
test_animal = {
    "name": "Test Dog",
    "animal_type": "Dog",
    "breed": "Labrador Retriever",
    "age_upon_outcome": "2 years",
    "color": "Black",
    "outcome_type": "Adoption"
}

create_result = shelter.create(test_animal)
print(f"\nCREATE Test:")
print(f"Insert successful: {create_result}")

# Test READ - Use your read function to return records from the aac database
print(f"\nREAD Test:")

# Read back the record we just created
read_result = shelter.read({"name": "Test Dog"})
print(f"Found {len(read_result)} record(s) with name 'Test Dog':")
for animal in read_result:
    print(animal)

# Also test reading existing data from the database
print(f"\nREAD Test (existing data):")
existing_dogs = shelter.read({"animal_type": "Dog", "breed": "Labrador Retriever"})
print(f"Found {len(existing_dogs)} Labrador Retriever(s) in database")
if existing_dogs:
    print("First result:", existing_dogs[0])
```

```python
# Test UPDATE - Modify a record in the aac database
print("\n" + "="*50)
print("UPDATE Test:")
print("="*50)

# Update the Test Dog's age
update_result = shelter.update(
    {"name": "Test Dog"},  # Find records with this name
    {"$set": {"age_upon_outcome": "3 years"}}  # Update the age to 3 years
)
print(f"Modified {update_result} record(s)")

# Verify the update worked
print("\nVerifying update:")
verify_update = shelter.read({"name": "Test Dog"})
print(f"Found {len(verify_update)} record(s) with name 'Test Dog'")
for animal in verify_update:
    print(f"Updated age_upon_outcome: {animal['age_upon_outcome']}")
```

```python
# Test DELETE - Remove the test record from the database
print("\n" + "="*50)
print("DELETE Test:")
print("="*50)

# Delete all records with name "Test Dog"
delete_result = shelter.delete({"name": "Test Dog"})
print(f"Deleted {delete_result} record(s)")

# Verify the delete worked
print("\nVerifying delete:")
verify_delete = shelter.read({"name": "Test Dog"})
print(f"Found {len(verify_delete)} record(s) with name 'Test Dog' (should be 0)")

if len(verify_delete) == 0:
    print("✓ DELETE operation successful - all Test Dog records removed")
else:
    print("✗ DELETE operation incomplete - some records still exist")
```
